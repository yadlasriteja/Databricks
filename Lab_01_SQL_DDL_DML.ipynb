```sql

SELECT
    current_catalog() AS current_catalog,
    current_schema() AS current_schema;
```

```sql

CREATE SCHEMA IF NOT EXISTS workspace.lab_db;

CREATE OR REPLACE TABLE workspace.lab_db.customers (
    customer_id INT,
    name STRING,
    email STRING,
    signup_date DATE,
    status STRING
)
USING DELTA;
```

```sql

INSERT INTO workspace.lab_db.customers
VALUES
    (101, 'Alice Smith', 'alice@example.com',
     DATE '2024-01-15', 'Active'),
    (102, 'Bob Jones', 'bob@example.com',
     DATE '2024-02-01', 'Pending');
```

```sql

SELECT *
FROM workspace.lab_db.customers
ORDER BY customer_id;
```

```sql

UPDATE workspace.lab_db.customers
SET email = 'alice.smith@example.com'
WHERE customer_id = 101;

SELECT *
FROM workspace.lab_db.customers
WHERE customer_id = 101;
```

```sql

INSERT INTO workspace.lab_db.customers
VALUES
    (999, 'Temporary User', 'temp@example.com',
     DATE '2024-03-01', 'Testing');

DELETE FROM workspace.lab_db.customers
WHERE customer_id = 999;

SELECT *
FROM workspace.lab_db.customers
ORDER BY customer_id;
```

```sql

MERGE INTO workspace.lab_db.customers AS target
USING (
    SELECT
        102 AS customer_id,
        'Bob Jones' AS name,
        'bob_new@example.com' AS email,
        DATE '2024-02-01' AS signup_date,
        'Active' AS status
) AS source
ON target.customer_id = source.customer_id

WHEN MATCHED THEN
    UPDATE SET
        target.email = source.email,
        target.status = source.status

WHEN NOT MATCHED THEN
    INSERT (
        customer_id, name, email, signup_date, status
    )
    VALUES (
        source.customer_id,
        source.name,
        source.email,
        source.signup_date,
        source.status
    );
```

```sql

SELECT *
FROM workspace.lab_db.customers
ORDER BY customer_id;
```
